#GENERATORS

**Regression**

```python
from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=500,n_features=5, n_targets=1, shuffle=True,noise=0.1, random_state=42)
print("Fm:",X.shape)
print("Lv:",y.shape)
```

```python
X[:5]
```

**Classification**

```python
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=500,n_features=5, n_classes=2, n_informative=2, n_redundant=0, random_state=42)
print("Fm:",X.shape)
print("Lv:",y.shape)
```
